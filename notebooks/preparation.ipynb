```python
import pandas as pd
import numpy as np
from pathlib import Path

url_VinRouge = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
url_VinBlanc = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv"
vin_blanc = pd.read_csv(url_VinBlanc, sep = ";")
vin_rouge= pd.read_csv(url_VinRouge, sep = ";")

vin_rouge['type'] = "rouge"
vin_blanc['type'] = "blanc"
vin = pd.concat([vin_rouge, vin_blanc], ignore_index=True)

vin.head(10)
```
